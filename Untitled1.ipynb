```python
def plotar_linha(df, x, y, hue, titulo, xlabel, ylabel):
    """
    Cria um gráfico de linhas para visualizar a evolução de uma variável ao longo do tempo.
    """
    plt.figure(figsize=(12, 7))
    sns.lineplot(data=df, x=x, y=y, hue=hue, marker='o')
    plt.title(titulo, fontsize=16)
    plt.xlabel(xlabel, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    if hue is not None:
        plt.legend(title=hue.capitalize(), bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True)
    plt.show()

def plotar_barra(df, x, y, titulo, xlabel, ylabel, orient='v'):
    """
    Cria um gráfico de barras para comparar categorias.
    """
    plt.figure(figsize=(12, 7))
    sns.barplot(data=df, x=x, y=y, palette='viridis', orient=orient)
    plt.title(titulo, fontsize=16)
    plt.xlabel(xlabel, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    plt.show()

def plotar_dispersao(df, x, y, size, hue, titulo, xlabel, ylabel):
    """
    Cria um gráfico de dispersão (bolhas) para analisar a relação entre três variáveis numéricas.
    """
    plt.figure(figsize=(14, 8))
    sns.scatterplot(data=df, x=x, y=y, size=size, hue=hue, sizes=(20, 1000), alpha=0.7, palette='muted')
    plt.title(titulo, fontsize=16)
    plt.xlabel(xlabel, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    plt.xscale('log') # Escala logarítmica é comum para PIB per capita
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True)
    plt.show()

print("Funções de visualização criadas com sucesso.")
```

```python
# Calculando a média global por ano
df_global_ano = df.groupby('year').agg({
    'lifeExp': 'mean',
    'gdpPercap': 'mean',
    'pop': 'sum'
}).reset_index()

# Plotando a evolução da expectativa de vida
plotar_linha(df_global_ano, 'year', 'lifeExp', None,
             'Evolução da Expectativa de Vida Média Global (1952-2007)',
             'Ano', 'Expectativa de Vida (Média)')
```

```python
# Usando o mesmo DataFrame agrupado da análise anterior
plotar_linha(df_global_ano, 'year', 'gdpPercap', None,
             'Evolução do PIB per Capita Médio Global (1952-2007)',
             'Ano', 'PIB per Capita (Média)')
```

```python
plotar_linha(df_global_ano, 'year', 'pop', None,
             'Crescimento da População Mundial (1952-2007)',
             'Ano', 'População Total (em bilhões)')
```

```python
# Calculando a média por continente e ano
df_continente_ano = df.groupby(['continent', 'year'])['lifeExp'].mean().reset_index()

plotar_linha(df_continente_ano, 'year', 'lifeExp', 'continent',
             'Evolução da Expectativa de Vida por Continente',
             'Ano', 'Expectativa de Vida (Média)')
```

```python
df_pib_continente_ano = df.groupby(['continent', 'year'])['gdpPercap'].mean().reset_index()

plotar_linha(df_pib_continente_ano, 'year', 'gdpPercap', 'continent',
             'Evolução do PIB per Capita por Continente',
             'Ano', 'PIB per Capita (Média)')
```

```python
df_2007 = df[df['year'] == 2007]

plotar_dispersao(df_2007, 'gdpPercap', 'lifeExp', 'pop', 'continent',
                 'PIB per Capita vs. Expectativa de Vida em 2007',
                 'PIB per Capita (log)', 'Expectativa de Vida')
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=df_2007, x='lifeExp', kde=True, bins=25)
plt.title('Distribuição da Expectativa de Vida Mundial em 2007', fontsize=16)
plt.xlabel('Expectativa de Vida', fontsize=12)
plt.ylabel('Frequência (Nº de Países)', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(data=df_2007, x='gdpPercap', kde=True, bins=30, log_scale=True)
plt.title('Distribuição do PIB per Capita Mundial em 2007 (Escala Log)', fontsize=16)
plt.xlabel('PIB per Capita (log)', fontsize=12)
plt.ylabel('Frequência (Nº de Países)', fontsize=12)
plt.show()
```

```python
bottom_10_lifeexp = df_2007.sort_values('lifeExp', ascending=True).head(10)

plotar_barra(bottom_10_lifeexp, 'lifeExp', 'country',
             '10 Países com Menor Expectativa de Vida em 2007',
             'Expectativa de Vida', 'País', orient='h')
```

```python
top_10_gdp = df_2007.sort_values('gdpPercap', ascending=False).head(10)

plotar_barra(top_10_gdp, 'gdpPercap', 'country',
             'Top 10 Países por PIB per Capita em 2007',
             'PIB per Capita', 'País', orient='h')
```

```python
top_10_pop = df_2007.sort_values('pop', ascending=False).head(10)

plotar_barra(top_10_pop, 'pop', 'country',
             'Top 10 Países Mais Populosos em 2007',
             'População (em bilhões)', 'País', orient='h')
```
